```python
from Shoe import shoe, stack, scripts
d = 6
bet = 5
ss = 200

stack = stack(ss)
stack.entbet(5)
shoe1 = shoe(d)
shoe1.shuffle()
```

```python
shoe1.deal()

print(f'You Have {shoe1.cards}')
print(f'Dealer Shows {shoe1.dealerup}')
```

```python
shoe1.dealerup = 'A'
shoe1.dealerdown = '5'
shoe1.dlist = ['A', '5']
shoe1.order.append('10')
print(f'Dealer Shows {shoe1.dealerup}')



```

```python
shoe1.bjcheck()

if shoe1.playerbj and shoe1.dealerbj:
    print('Two Blackjacks, Push')
    stack.push()
    print(f'You have {stack.size} chips')

elif shoe1.playerbj:
    print('Blackjack!')
    stack.win()
    print(f'You have {stack.size} chips')

elif shoe1.dealerbj:
    print(f'Dealer Shows {shoe1.dealerdown}')
    print('FUCK YOU')
    stack.lose()
    print(f'You have {stack.size} chips')

else:
    print('No Blackjack')
```

```python
shoe1.splcheck()
nose = []
while shoe1.spavailable:

    shoe1.spvloc = [x for x in shoe1.spvloc if x not in nose]
    if shoe1.spvloc == []:
        break

    for i in shoe1.spvloc:
        item = shoe1.cards[i]

        spl = input(f"Would you like to split hand {i}: {item}? (y/n)")

        if spl == 'y':
            shoe1.split(i)
            print(f'Your Cards Are {shoe1.cards}')
        elif spl == 'n':
            nose.append(i)
        else:
            print("Error, Return (y/n)")

    shoe1.splcheck()



print('No Splits Available')
```

```python
for i in range(len(shoe1.cards)):
    item = shoe1.cards[i]
    dbl = input("Would you like to double? (y/n)")
    if dbl == "y":
        print("Your bet has been doubled!")
        shoe1.double(i)
        print(f"Your Hand is {shoe1.cards[i]}")
        shoe1.valchk(i)
        if shoe1.value > 21:
            print("BUST")     
            stack.lose()
            stack.lose()
            print(f'You have {stack.size} chips')
            del shoe1.cards[i]
    elif dbl == "n":
        continue
    else:
        print("Please Enter (y/n)")
```

```python
print(f"Dealer is Showing {shoe1.dealerup}")


for i in range(len(shoe1.cards)):
    item = shoe1.cards[i]

    while 1 == 1:
        ans = input(f"Hit or Stand on hand {i}: {item}")
        if ans.lower() == "hit":
            shoe1.hit(i)
            print(f'Your Cards Are {shoe1.cards}')
            shoe1.valchk(i)
            if shoe1.value > 21:
                print("BUST (rip bozo)")     
                stack.lose()
                print(f'You have {stack.size} chips')
                del shoe1.cards[i]
                break
        elif ans.lower() == "stand":
            shoe1.valchk(i)        
            break
        else:
            print("Please Enter Either Hit or Stand.")

        

    
```

```python
print(f'Dealer shows {shoe1.dlist}')

shoe1.dchk()

print(f'Dealer now has {shoe1.dlist}')

if shoe1.dvalue > 21:
    print("Dealer Busts!")
    for i in range(len(shoe1.cards)):
        stack.win()
    print(f"You now have {stack.size} chips")
else:
    for i in range(len(shoe1.cards)):
        item = shoe1.cards[i]
        shoe1.valchk(i)
        if shoe1.value > shoe1.dvalue:
            print(f"You win with {item}")
            stack.win()
        elif shoe1.value == shoe1.dvalue:
            print(f"Push with {item}")
        else:
            print(f"You lose with {item}")
            stack.lose()

```
